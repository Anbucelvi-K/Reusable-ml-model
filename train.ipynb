```python
import joblib
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, classification_report
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline
```

```python
# Step 1: Load the 20 Newsgroups dataset (train and test sets)
print("Loading 20 Newsgroups dataset...")

train = fetch_20newsgroups(
    subset='train',
    remove=('headers', 'footers', 'quotes')
)

test = fetch_20newsgroups(
    subset='test',
    remove=('headers', 'footers', 'quotes')
)
```

```python
# Step 2: Create a pipeline with TF-IDF and Naive Bayes
model = make_pipeline(
    TfidfVectorizer(),
    MultinomialNB()
)
```

```python
# Step 3: Train the model
print("Training the model...")

model.fit(train.data, train.target)

print("Training completed!")
```

```python
# Step 4: Make predictions on the test data
print("Testing the model...")

predictions = model.predict(test.data)

print("Testing completed!")
```

```python
# Step 5: Evaluate the model
accuracy = accuracy_score(test.target, predictions)

print(f"Accuracy: {accuracy:.2f}")
```

```python
# Step 6: Display the classification report
print("\nClassification Report:")
print(classification_report(test.target, predictions))
```

```python
# Step 7: Save the model to disk
model_filename = '20newsgroups_model.joblib'

joblib.dump(model, model_filename)

print(f"Model saved successfully as {model_filename}")
```
